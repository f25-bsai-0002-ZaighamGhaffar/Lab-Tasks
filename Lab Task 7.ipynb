```python
#Task 1:
class student:
    def __init__(self,name,age,course):
        self.name=name
        self.age=age
        self.course=course
    def display(self):
        print("Name: ",self.name)
        print("Age: ",self.age)
        print("Course: ",self.course)
 
s1=student("Ali",23,"Python")  
s1.display()
s2=student("Zain",12,"AI")    
s2.display()
```

```python
#Task 2:
class student:
    def __init__(self,name,age,course):
        self.name=name
        self.age=age
        self.course=course
    def display(self):
        print("Name: ",self.name)
        print("Age: ",self.age)
        print("Course: ",self.course)
class graudate(student):
    def __init__(self,name,age,course,research):
        super().__init__(name,age,course)
        self.research=research
    def display(self):    
        super().display()
        print("Research: ",self.research)    
g1=graudate("Ali",23,"Python","Palestine")  
g1.display()
g2=graudate("Zain",12,"AI","Genz")    
g2.display()
```
